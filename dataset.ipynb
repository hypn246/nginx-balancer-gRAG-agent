```python
import os
import pandas as pd
import numpy as np
```

```python
Dataset source: https://www.kaggle.com/datasets/sandhyapeesara/cloud-anomaly-data
```

```python
path='../data/dataset/Cloud_Anomaly_Dataset.csv'
```

```python
data=pd.read_csv(path)
print(data)
```

```python
clean_data=data.dropna()
clean_data = clean_data[clean_data["cpu_usage"] >= 5]
clean_data = clean_data[clean_data["memory_usage"] >= 5]
print(clean_data)
```

```python
#x.x.x.0 & x.x.x.255 skip cuz net and boardcast address
def assign_IP(n):
    ips = []

    a, b, c, d = 10, 0, 0, 1
    for i in range(n):
        if(d==256):
            d=1
            c=c+1
            if(c==256):
                c=0
                b=b+1
                if(b==256):
                    b=0
                    a=a+1
                    if(a==256):
                        if(a==b==c==d==255):
                            print(f"out of IP address. Stop at {i} rows")
                            break
        ips.append(f"{a}.{b}.{c}.{d}")
        d=d+1
    return ips
```

```python
#clean_data.drop(["ip"], axis=1, inplace=True)
```

```python
print(clean_data)
```

```python
ips = assign_IP(len(clean_data))
clean_data.insert(1, "ip", ips)
```

```python
clean_data.tail()
```

```python
final_data=clean_data.drop(["vm_id","timestamp","power_consumption","num_executed_instructions","energy_efficiency","task_type","task_priority","task_status","Anomaly status"], axis=1)
```

```python
print(final_data)
```

```python
latency=(
    (clean_data["execution_time"]/100) *
    ((clean_data["cpu_usage"] + clean_data["memory_usage"]) /200) *  clean_data["network_traffic"]
)
print(latency)
```

```python
final_data.insert(len(final_data.columns),"latency",latency)
final_data[["latency", "cpu_usage", "memory_usage"]] =final_data[["latency", "cpu_usage", "memory_usage"]].round(2)
final_data[["network_traffic","execution_time"]] =final_data[["network_traffic", "execution_time"]].round(0)
```
